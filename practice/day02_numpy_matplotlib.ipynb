```python
#-----functions----

def average(numbers):
    total = 0
    for n in numbers:
        total = total + n
    return total / len(numbers)

print(average([0.80, 0.82, 0.78, 0.82])) 

def heart_rate(gap_seconds):
    return 60 / gap_seconds
print(heart_rate(0.8))


```

```python
# --- sine wave exercise --
import numpy as np
import matplotlib.pyplot as plt
fs = 200                       # sampling rate: 200 measurements per second
t = np.arange(0, 5, 1/fs)      # time: 0, 0.005, 0.010 ... seconds
x = np.sin(2*np.pi*1*t)        # a 1 Hz wave
print(x.shape, x[:5])          # how many numbers, first five
print(x.mean(), x.std())       # average and spread
print(np.diff(x)[:3])          # differences between neighbors
plt.plot(t, x); plt.xlabel("seconds"); plt.show()
```

```python
#----exercise further -----#

import numpy as np
import matplotlib.pyplot as plt

fs = 200                       # sampling rate: 200 measurements per second
t = np.arange(0, 5, 1/fs)      # time: 0, 0.005, 0.010 ... seconds
x2 = np.sin(2*np.pi*2*t)        # a 2 Hz wave
x5 = np.sin(2*np.pi*5*t)        # a 5 Hz wave
plt.plot(t,x2,label="2 Hz wave",color="red")
plt.plot(t,x5,label="5 Hz wave",color="blue")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.title("Sine Waves")
plt.legend()
plt.show()
```

```python
def mean_and_std(a) :
    a = np.array(a)
    return a.mean(), a.std()

print(mean_and_std([0.80, 0.82, 0.78, 0.82]))
    
```
